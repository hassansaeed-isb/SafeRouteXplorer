```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, cohen_kappa_score
from scipy.stats import pearsonr, spearmanr

# Read the human ratings (assuming the column is '3')
human = pd.read_csv("hr.csv")
# Rename the human rating column to "Human"
human.columns = ["Human"]

# Read model prediction files and store them in a dictionary
models = {
    "Llama8B": pd.read_csv("Extracted_Severity_Data_-_Llama8B_File.csv"),
    "Gemma2": pd.read_csv("Extracted_Severity_Data_-_Gemma2_File.csv"),
    "Mixral": pd.read_csv("Extracted_Severity_Data_-_Mixral_File.csv"),
    "Deepseek": pd.read_csv("Extracted_Severity_Deepseek.csv"),
    "Llama3.3": pd.read_csv("Extracted_Severity_Data_llama3.3.csv")
}

# We'll assume the predictions column is named "Severity" in all model files.
# Combine human ratings with each model predictions assuming they are row aligned.
results = {}

for model_name, df in models.items():
    # Make sure the predictions are numeric
    predictions = df["Severity"].astype(float)
    humans = human["Human"].astype(float)

    # Calculate MAE and RMSE
    mae = mean_absolute_error(humans, predictions)
    rmse = np.sqrt(mean_squared_error(humans, predictions))

    # Calculate Pearson and Spearman correlations
    pearson_corr, _ = pearsonr(humans, predictions)
    spearman_corr, _ = spearmanr(humans, predictions)

    # Calculate Quadratic Weighted Kappa
    # Since cohen_kappa_score expects integer labels, we round if necessary.
    qwk = cohen_kappa_score(humans.round().astype(int), predictions.round().astype(int), weights='quadratic')

    results[model_name] = {
        "MAE": mae,
        "RMSE": rmse,
        "Pearson": pearson_corr,
        "Spearman": spearman_corr,
        "Quadratic Weighted Kappa": qwk
    }

# Display the results
results_df = pd.DataFrame(results).T
print(results_df)
```

```python
import pandas as pd
from transformers import pipeline

# Define the model name from Hugging Face
MODEL = "jy46604790/Fake-News-Bert-Detect"

# Load the fake news detection pipeline
clf = pipeline("text-classification", model=MODEL, tokenizer=MODEL)

# Load the hazardous road incidents data
df = pd.read_csv("Hazardous_Road_Incidents.csv")

# Function to classify a single tweet using the fake news classifier
def classify_tweet(tweet_text):
    try:
        result = clf(tweet_text)
        label = result[0]['label']
        score = result[0]['score']
    except Exception as e:
        label = "ERROR"
        score = 0.0
    return pd.Series([label, score])

# Apply the classifier to each tweet (now using the correct column "text")
df[['fake_news_label', 'fake_news_score']] = df['text'].apply(classify_tweet)

# Save the updated DataFrame to a new CSV file
df.to_csv("Hazardous_Road_Incidents_with_Fake_News.csv", index=False)

print("Fake news classification completed and results saved to 'Hazardous_Road_Incidents_with_Fake_News.csv'.")
```

```python
import pandas as pd
from transformers import pipeline

# Define the model name from Hugging Face
MODEL = "jy46604790/Fake-News-Bert-Detect"

# Initialize the text classification pipeline
clf = pipeline("text-classification", model=MODEL, tokenizer=MODEL)

# Load your hazardous road incidents tweets.
# Assumes the CSV has a column named 'tweet' containing the tweet texts.
df = pd.read_csv("Hazardous_Road_Incidents.csv")

# Function to classify each tweet using the fake news classifier.
# The logic: if the model indicates "fake" (i.e. returns "LABEL_0")
# and its confidence is above 0.30, we mark the tweet as fake (label 1);
# otherwise, we mark it as true (label 0).
def classify_tweet(tweet_text, threshold=0.30):
    try:
        result = clf(tweet_text)
        # Initialize fake confidence to 0.0
        fake_confidence = 0.0
        # The model returns "LABEL_0" for fake news, "LABEL_1" for real news.
        # We consider the fake confidence only if the model predicts "LABEL_0".
        if result[0]['label'] == "LABEL_0":
            fake_confidence = result[0]['score']
        # Apply threshold: if fake confidence > threshold, mark as fake (1),
        # otherwise, mark as true (0).
        if fake_confidence > threshold:
            label = 1  # Fake
        else:
            label = 0  # True
    except Exception as e:
        label = "ERROR"
        fake_confidence = 0.0
    return pd.Series([label, fake_confidence])

# Apply the classifier to each tweet.
# This creates two new columns: 'predicted_label' and 'fake_confidence'.
df[['predicted_label', 'fake_confidence']] = df['text'].apply(lambda x: classify_tweet(x, threshold=0.30))

# Save the updated DataFrame to a new CSV file.
df.to_csv("Hazardous_Road_Incidents_with_Adjusted_Labels.csv", index=False)

print("Classification completed and results saved to 'Hazardous_Road_Incidents_with_Adjusted_Labels.csv'.")
```
